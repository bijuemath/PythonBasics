# Web Frameworks: FastAPI and Streamlit
This notebook provides the foundation for building modern web applications with Python.

## 1. FastAPI (The Backend)
FastAPI is a modern, fast web framework for building APIs with Python.

```python
# This code is meant to be saved in a file like 'main.py'
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "FastAPI is running!"}

print("Run this server using: uvicorn main:app --reload")
```

## 2. Streamlit (The Frontend)
Streamlit lets you create interactive web apps for data science and AI projects.

```python
# This code is meant to be saved in a file like 'app.py'
import streamlit as st

st.title("Streamlit UI")
st.write("Hello! This is a simple UI.")

if st.button("Click Me"):
    st.success("Button Clicked!")

print("Run this UI using: streamlit run app.py")
```
