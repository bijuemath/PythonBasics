# Prompt Engineering: System Message, Few-Shot, and CoT
This notebook provides functional examples and explanations for the three core pillars of modern prompt engineering.

## 1. System Message (Defining Persona)
The **System Message** acts as the foundation for the AI's personality. In the example below, we simulate setting a persona that is strictly technical.

```python
# Simulation of a System Role definition
system_message = """You are a Senior Security Engineer. 
Your responses must be technical, objective, and highlight potential vulnerabilities."""

user_query = "How do I secure a web server?"

print(f"SYSTEM ROLE: {system_message}")
print(f"USER: {user_query}")
```

## 2. Few-Shot Prompting (Pattern Matching)
By providing a few 'shots' (examples), we train the model on the specific format we want.

```python
# Example of 3-shot prompting for classification
few_shot_prompt = """
Classify the sentiment of these movie reviews:

Review: The plot was predictable and boring.
Sentiment: Negative

Review: An absolute masterpiece of modern cinema!
Sentiment: Positive

Review: It was okay, but I wouldn't watch it again.
Sentiment: Neutral

Review: I haven't laughed this hard in years.
Sentiment:
"""

print(few_shot_prompt)
```

## 3. Chain of Thought (CoT)
CoT is essential for logic. It prevents the AI from 'guessing' and encourages internal verification.

```python
# Forcing logic with 'Let\'s think step by step'
cot_prompt = """
Question: A farmer has 15 sheep. All but 8 die. How many are left?

Reasoning: Let's think step by step.
1. The farmer starts with 15 sheep.
2. The phrase 'all but 8 die' means that 8 sheep survived.
3. Therefore, those 8 are the only ones left.
Answer: 8
"""

print(cot_prompt)
```
