# Python Basics Part 2: Data Structures & Loops
Now that you know variables and conditions, this notebook covers how to store groups of data and repeat actions efficiently.

## 1. Lists
Lists are used to store multiple items in a single variable. They are ordered, changeable, and allow duplicate values.

```python
fruits = ["apple", "banana", "cherry"]
print(f"Original List: {fruits}")

# Accessing items by index (starts at 0)
print(f"First fruit: {fruits[0]}")

# Adding an item
fruits.append("orange")

# Changing an item
fruits[1] = "blueberry"

print(f"Modified List: {fruits}")
print(f"Length of list: {len(fruits)}")
```

## 2. While Loops
A `while` loop repeats a block of code as long as a specified condition is true.

```python
count = 1
while count <= 5:
    print(f"Count is: {count}")
    count += 1  # Important: increment to avoid infinite loop
```

## 3. For Loops
A `for` loop is used for iterating over a sequence (like a list or a range of numbers).

```python
print("Iterating through a list:")
colors = ["red", "green", "blue"]
for color in colors:
    print(color)

print("\nUsing range(start, stop):")
for i in range(1, 4):
    print(f"Number {i}")
```

## 4. Dictionaries
Dictionaries store data values in **key:value** pairs. They are great for representing real-world objects.

```python
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

print(f"Car Brand: {car['brand']}")

# Adding a new key-value pair
car["color"] = "red"

print(car)
```

## 5. Functions
Functions are blocks of code that only run when called. They help make your code reusable.

```python
def greet_user(username):
    """Displays a simple greeting."""
    return f"Hello, {username}! Welcome to Python."

message = greet_user("Student")
print(message)
```
