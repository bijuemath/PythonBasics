# Python Basics: Getting Started
This notebook covers the fundamental building blocks of Python programming, from your first line of code to basic logic control.

## 1. Hello World
The `print()` function is used to output text to the console.

```python
# This is your first line of code
print("Hello, World!")
```

## 2. Variables and Literals
Variables are containers for storing data values. Python has various literals (data types) like integers, floats, and strings.

```python
# Integer literal
age = 25 

# Float (decimal) literal
price = 19.99 

# String literal
name = "Alice" 

# Boolean literal
is_python_fun = True

print(f"Name: {name}, Age: {age}, Price: {price}")
```

## 3. Basic Arithmetic
Python can be used as a calculator for simple or complex math.

```python
a = 10
b = 3

print("Addition:", a + b)
print("Division:", a / b)
print("Floor Division:", a // b) # Removes decimals
print("Remainder (Modulus):", a % b)
```

## 4. Input and Type Conversion
You can take user input using `input()`. Note that input is always received as a string, so you may need to convert it.

```python
# user_age = input("Enter your age: ") # Uncomment this to run locally
user_age = "20" # Example value

# Convert string to integer to do math
years_to_100 = 100 - int(user_age)
print(f"You will be 100 in {years_to_100} years!")
```

## 5. If...Else Conditions
Conditional statements allow you to run code only if a specific condition is met. **Note:** Indentation (the space at the start of the line) is required in Python.

```python
temperature = 25

if temperature > 30:
    print("It's a hot day!")
elif temperature >= 20:
    print("It's a pleasant day.")
else:
    print("It's a cold day.")
```

## 6. Logical Operators
You can combine conditions using `and`, `or`, and `not`.

```python
has_id = True
is_adult = True

if has_id and is_adult:
    print("Entry allowed")
else:
    print("Entry denied")
```
