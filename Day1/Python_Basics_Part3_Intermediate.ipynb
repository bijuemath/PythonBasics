This notebook covers essential topics for writing robust programs, including error handling, working with files, and using external modules.

## 1. Exception Handling (Try...Except)
To prevent your program from crashing when an error occurs, you can use `try` and `except` blocks.

```python
try:
    number = int(input("Enter a number to divide 10 by: "))
    result = 10 / number
    print(f"Result is {result}")
except ValueError:
    print("Invalid input! Please enter a whole number.")
except ZeroDivisionError:
    print("You cannot divide by zero!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

## 2. List Comprehensions
A concise way to create lists based on existing lists.

```python
numbers = [1, 2, 3, 4, 5]
# Create a list of squares
squares = [n**2 for n in numbers]
print(f"Numbers: {numbers}")
print(f"Squares: {squares}")
```

## 3. Working with Modules
Python comes with many built-in modules. You use the `import` keyword to use them.

```python
import math
import random

print(f"Square root of 16 is: {math.sqrt(16)}")
print(f"Random number between 1 and 100: {random.randint(1, 100)}")
```

## 4. File I/O (Reading and Writing Files)
You can interact with files on your computer using the `with` statement.

```python
# Writing to a file
with open("test.txt", "w") as file:
    file.write("This is a test file created by Python!\nLearning VS Code is fun.")

# Reading from a file
with open("test.txt", "r") as file:
    content = file.read()
    print("File Content:")
    print(content)
```

## 5. Tuples and Sets
Other useful data collections:
- **Tuples:** Immutable (cannot be changed) lists.
- **Sets:** Unordered collections with no duplicate items.

```python
my_tuple = (10, 20, 30)
my_set = {1, 2, 2, 3, 4, 4} # Duplicates are removed

print(f"Tuple: {my_tuple}")
print(f"Set (unique items only): {my_set}")
```
